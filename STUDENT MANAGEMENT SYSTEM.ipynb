```python
class Student:
    def __init__(self, name, roll_no, grade, marks):
        self.name = name
        self.__roll_no = roll_no     # private variable
        self.__grade = grade         # private variable
        self.__marks = marks         # private variable

    # Getter methods (read-only access)
    def get_roll_no(self):
        return self.__roll_no

    def get_grade(self):
        return self.__grade

    def get_marks(self):
        return self.__marks

    # Setter methods (controlled modification)
    def set_grade(self, grade):
        self.__grade = grade

    def set_marks(self, marks):
        self.__marks = marks

    # Display student info
    def display_info(self):
        print(f"🎓 Name: {self.name}")
        print(f"🔢 Roll No: {self.__roll_no}")
        print(f"🏷️ Grade: {self.__grade}")
        print(f"📊 Marks: {self.__marks}")
        print("-" * 35)


class StudentManager:
    def __init__(self):
        self.students = []

    # Add new student
    def add_student(self):
        name = input("Enter Student Name: ")
        roll_no = input("Enter Roll Number: ")
        grade = input("Enter Grade: ")
        marks = float(input("Enter Marks: "))

        student = Student(name, roll_no, grade, marks)
        self.students.append(student)
        print("✅ Student added successfully!")

    # View all students
    def view_students(self):
        if not self.students:
            print("⚠️ No students available.")
        else:
            print("\n📋 Student List:")
            for s in self.students:
                s.display_info()

    # Search student by roll no
    def search_student(self):
        roll = input("Enter Roll Number to search: ")
        for s in self.students:
            if s.get_roll_no() == roll:
                print("\n🔍 Student Found:")
                s.display_info()
                return
        print("❌ Student not found!")

    # Delete student by roll no
    def delete_student(self):
        roll = input("Enter Roll Number to delete: ")
        for s in self.students:
            if s.get_roll_no() == roll:
                self.students.remove(s)
                print("🗑️ Student deleted successfully!")
                return
        print("❌ Student not found!")

    # Update marks (with encapsulation)
    def update_marks(self):
        roll = input("Enter Roll Number to update marks: ")
        for s in self.students:
            if s.get_roll_no() == roll:
                new_marks = float(input("Enter new marks: "))
                s.set_marks(new_marks)
                print("✅ Marks updated successfully!")
                return
        print("❌ Student not found!")


# --- Main Program ---
def main():
    print("🎓 Welcome to Student Management System (Encapsulation Demo)\n")
    manager = StudentManager()

    while True:
        print("\n1. Add Student\n2. View Students\n3. Search Student\n4. Delete Student\n5. Update Marks\n6. Exit")
        choice = input("Enter your choice: ")

        if choice == '1':
            manager.add_student()
        elif choice == '2':
            manager.view_students()
        elif choice == '3':
            manager.search_student()
        elif choice == '4':
            manager.delete_student()
        elif choice == '5':
            manager.update_marks()
        elif choice == '6':
            print("👋 Exiting system. Goodbye!")
            break
        else:
            print("❌ Invalid choice! Try again.")


if __name__ == "__main__":
    main()
```
